```python
from recipes.rag.rag_service import RAGService
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
rag_service = RAGService(embedding_model="text-embedding-3-small", query_model="gpt-4o", db_url=os.getenv("MONGO_DB_URL"), collection=os.getenv("MONGO_COLLECTION_NAME_RECIPES")) # type: ignore
```

```python
rag_service.setup()
```

```python
# Hier Anfrage festlegen

anfrage = "Ich habe eine Party und möchte etwas leckeres zum Naschen anbieten. Ich habe aber nicht viel Zeit und keine Lust auf viel Aufwand. Ich habe Schokolade, Obst, Nüsse, Pudding"

# Schritt 1 - Retrieval: Die Top-3-Treffer der Vektorsuche anzeigen
# (das ist genau das, was die $vectorSearch-Pipeline aus MongoDB zurückliefert)
retrieved_recipes = rag_service.query_database(anfrage)
print(rag_service.format_retrieved_recipes(retrieved_recipes))
```

```python
# Schritt 2 - Generation: Antwort des LLM auf Basis der oben gefundenen Rezepte
antwort = rag_service.generate_response(anfrage, retrieved_recipes)
print(antwort)
```
